## VADER Sentiment Analyzer

VADER stands for Valence Aware Dictionary Entity Recognition. No preprocessing or training is required. It is rule based approach for sentiment analysis

```python
import nltk
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
sia = SentimentIntensityAnalyzer()
```

```python
review = "The movie was terrific to watch and absolute treat to watch!"
```

```python
sia.polarity_scores(review)  # Ouputs pos, neg, neu and compound score for the review based on VADER Lexicon.
```

If the compound score is:

* **> 0.05** - positive sentiment
* **< -0.05** - negative sentiment
* **-0.05 < score < 0.05** - neutral sentiment

## Reviews Preprocessing

By preprocessing the review we remove:

* punctuations
* numbers
* stopwords
* special characters (@, #, !, ...)
* excessive white spaces

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
import re
```

```python
def process_review(review):
    # Convert review into lowercase form
    review_lower = review.lower()

    # Remove punctuations, numbers
    review_puncs_nums_removed = re.sub(r"[^a-z\s]", "", review_lower)

    # Tokenize the review: convert sentence into words
    review_tokens = word_tokenize(review_puncs_nums_removed)

    # Remove stop words
    stop_words = stopwords.words('English')
    review_tokens_stopwords_removed = [token for token in review_tokens if token not in stop_words]

    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    lemmatized_tokens = [lemmatizer.lemmatize(token, pos='n') for token in review_tokens_stopwords_removed]

    # Put together all the tokens as a single string object
    return " ".join(lemmatized_tokens)
```

```python
process_review("The product was really good!!!! It made me happy :)")
```

```python
# Separating reviews and sentiment labels
processed_reviews = []
labels = []

for rev in reviews:
    doc, label = rev.strip().split('\t')
    processed_review = process_review(doc)
    processed_reviews.append(processed_review)
    labels.append(label)
```
